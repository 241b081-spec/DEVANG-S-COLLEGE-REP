# Minimum Cost Spanning Tree
Problem Description:
A spanning tree of a graph can be defined as a graph with minimal set of edges that connect all vertices. A minimum spanning tree of a graph is a spanning tree of the graph with least weight (where the weight is computed by adding the weights of all the edges in the spanning tree). In general, a graph can have multiple minimum spanning trees. The problem is to find a minimum spanning tree of a graph.

This implementation uses a dictionary to represent the weighted graph, where the vertices are keys, and the values are dictionaries containing their adjacent vertices and their weights.

Please note that this implementation assumes that the input graph is connected and weighted.

First, let's create a simple graph to demonstrate Prim's algorithm:

```python
graph = {
    "A": {"B": 10, "C": 15, "D": 20},
    "B": {"A": 10, "C": 12},
    "C": {"A": 15, "B": 12, "D": 18},
    "D": {"A": 20, "C": 18},
}
```

Now, let's implement Prim's algorithm:

```python
def prim_algorithm(graph):
    min_tree = {}
    visited = set()
    min_tree[list(graph.keys())[0]] = []
    min_weight = 0

    while len(visited) < len(graph):
        min_edge = float('inf')
        next_vertex = None
        for vertex in min_tree:
            for adj_vertex, weight in graph[vertex].items():
                if adj_vertex not in visited:
                    if weight < min_edge:
                        min_edge = weight
                        next_vertex = adj_vertex

        min_tree[next_vertex] = []
        visited.add(next_vertex)
        min_weight += min_edge

    return min_tree, min_weight
```

Now, let's run the Prim's algorithm on the example graph:
Keep in mind that this example uses a simple representation of a graph, but for more complex cases, you might want to use a library such as NetworkX to represent and manipulate graphs.

```python
mst, min_weight = prim_algorithm(graph)
print("Minimum Spanning Tree: ", mst)
print("Minimum Weight: ", min_weight)
```

```python
# your sample input of graph
from collections import deque

V, E = 6, 8
edges = [
    (1,2), (1,3), (2,4), (2,5),
    (3,5), (3,6), (4,6), (5,6)
]

adj = [[] for _ in range(V+1)]
for u, v in edges:
    adj[u].append(v)

def DFS(start):
    visited = [False] * (V+1)
    result = []

    def dfs(node):
        visited[node] = True
        result.append(node)
        for neighbor in adj[node]:
            if not visited[neighbor]:
                dfs(neighbor)

    dfs(start)
    return result

def BFS(start):
    visited = [False] * (V+1)
    queue = deque([start])
    visited[start] = True
    result = []

    while queue:
        node = queue.popleft()
        result.append(node)
        for neighbor in adj[node]:
            if not visited[neighbor]:
                visited[neighbor] = True
                queue.append(neighbor)

    return result

print("DFS from node 1:", DFS(1))  # Expected: [1, 2, 4, 6, 5, 3]
print("BFS from node 1:", BFS(1))  # Expected: [1, 2, 3, 4, 5, 6]
```

```python
#run prims algorithm for sample input

from collections import defaultdict
import heapq
import sys
weighted_edges = [
    (1,2,4), (1,3,8), (2,4,2), (2,5,11),
    (3,5,4), (3,6,3), (4,6,7), (5,6,2)
]

adj_weighted = defaultdict(list)
for u, v, w in weighted_edges:
    adj_weighted[u].append((v, w))
    adj_weighted[v].append((u, w))

def prims_mst(start):
    V = 6
    INF = sys.maxsize

    min_dist = [INF] * (V + 1)
    parent = [-1] * (V + 1)
    visited = [False] * (V + 1)

    min_heap = [(0, start)]
    min_dist[start] = 0

    total_cost = 0
    mst_edges = []

    while min_heap:
        dist, u = heapq.heappop(min_heap)

        if visited[u]:
            continue

        visited[u] = True
        total_cost += dist

        if parent[u] != -1:
            mst_edges.append((parent[u], u, dist))

        for v, weight in adj_weighted[u]:
            if not visited[v] and weight < min_dist[v]:
                min_dist[v] = weight
                parent[v] = u
                heapq.heappush(min_heap, (weight, v))

    return mst_edges, total_cost

mst_edges, total_cost = prims_mst(1)

print("Prim's MST Edges:")
for u, v, w in mst_edges:
    print(f"{u}-{v}: {w}")

print(f"\nTotal MST Cost: {total_cost}")
```

```python
# implement Kruskal's algorithm

from collections import defaultdict
weighted_edges = [
    (1,2,4), (1,3,8), (2,4,2), (2,5,11),
    (3,5,4), (3,6,3), (4,6,7), (5,6,2)
]

def find(parent, i):
    """Path compression in Union-Find"""
    if parent[i] != i:
        parent[i] = find(parent, parent[i])
    return parent[i]

def union(parent, rank, x, y):
    """Union by rank"""
    if rank[x] > rank[y]:
        parent[y] = x
    elif rank[x] < rank[y]:
        parent[x] = y
    else:
        parent[y] = x
        rank[x] += 1

def kruskals_mst():
    V = 6

    edges_sorted = sorted(weighted_edges, key=lambda x: x[2])

    parent = list(range(V+1))
    rank = [0] * (V+1)

    mst_edges = []
    total_cost = 0

    edges_used = 0

    print("Kruskal's Algorithm - Edge Selection:")
    for u, v, w in edges_sorted:
        x = find(parent, u)
        y = find(parent, v)

        print(f"Consider {u}-{v} (w={w}): roots {x} != {y} -> ", end="")

        if x != y:
            union(parent, rank, x, y)
            mst_edges.append((u, v, w))
            total_cost += w
            edges_used += 1
            print("ADDED")

            if edges_used == V-1:
                break
        else:
            print("CYCLE - SKIPPED")

    return mst_edges, total_cost

mst_edges, total_cost = kruskals_mst()

print("\n" + "="*40)
print("KRUSKAL'S MST RESULT:")
print("="*40)
for u, v, w in mst_edges:
    print(f"{u}-{v}: {w}")

print(f"\nTotal MST Cost: {total_cost}")
print("Number of edges in MST: 5")
```
